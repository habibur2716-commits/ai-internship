```python
def greet(name):
    print(f"Hello, {name}!")

greet("Ali")
```

```python
def add(a, b):
    return a + b

result = add(10, 5)
print(result)
```

```python
def greet_with_default(name="Guest"):
    print(f"Hello, {name}!")

greet_with_default()          # Hello, Guest!
greet_with_default("Sara")    # Hello, Sara!
```

```python
square = lambda x: x * x
print(square(6))
```
