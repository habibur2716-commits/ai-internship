```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(arr.mean())
print(arr.sum())
print(arr * 2)
```

```python
import pandas as pd

data = {
    "Name" : ["Habib", "Saim", "Minhaj"],
    "Age" : [21, 19, 21]
}

df = pd.DataFrame(data)
print(df)
print(df[df["Age"] > 27])
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Simple Line Chart")
plt.show()
```
